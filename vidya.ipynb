```python
n=int(input("enter the max number of candies"))
k=int(input("enter the min no of candies"))
C=1
r=n
while(C==1):
  m=int(input("Enter amount of candies needed:"))
  if(m>=n):
    print("invalid")
  else:
    r=r-m
    print("the candies sold:",m)
    if(r<=k):
        r=n
    print("sold=",m)
    print("Remaining=",r)
    C=int(input("DO you want to continue(1or0):"))

```

**duplic**

```python
n = int(input("Enter the number of elements: "))
a = []

for i in range(n):
    m = int(input("Enter the element: "))
    a.append(m)

print("Original list:", a)

result = []
for num in a:
    if num not in result:
        result.append(num)

print("List after removing duplicates:", result)
```

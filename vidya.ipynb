```python
n=int(input("enter the total no of monkey's"))
k=int(input("enter the no of banana a monkey can eat"))
j=int(input("enter the no of peanut a monkey can eat"))
m=int(input("enter the no of banana"))
p=int(input("enter the no of peanut"))
if(k==0 or j==0):
  print("invalid input")
else:
  n=n-(m//k+p//j)
  if(m%k!=0 or p%k!=0):
    n=n-1
  print("No of monkeys in tree:",n)


```

```python
n=int(input("enter the max number of candies"))
k=int(input("enter the min no of candies"))
C=1
r=n
while(C==1):
  m=int(input("Enter amount of candies needed:"))
  if(m>=n):
    print("invalid")
  else:
    r=r-m
    print("the candies sold:",m)
    if(r<=k):
        r=n
    print("sold=",m)
    print("Remaining=",r)
    C=int(input("DO you want to continue(1or0):"))

```

```python
n = int(input("Enter number of elements: "))
a = []
for i in range(n):
    a.append(int(input("Enter element: ")))
print("Reversed array:",a[::-1])
for i in range(n - 1, -1, -1):
    print(a[i], end=" ")
```

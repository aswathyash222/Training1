```python
n=int(input("enter the total no of monkey's"))
k=int(input("enter the no of banana a monkey can eat"))
j=int(input("enter the no of peanut a monkey can eat"))
m=int(input("enter the no of banana"))
p=int(input("enter the no of peanut"))
if(k==0 or j==0):
  print("invalid input")
else:
  n=n-(m//k+p//j)
  if(m%k!=0 or p%k!=0):
    n=n-1
  print("No of monkeys in tree:",n)


```
